```python
class Human:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def check_adult(self):
        if self.age >= 20:
            print(f"{self.name}さんは{self.age}歳です。20歳以上のため大人です。")
        else:
            print(f"{self.name}さんは{self.age}歳です。20歳未満のため大人ではありません。")

user1 = Human("Bob", 35)
user2 = Human("John", 18)
user3 = Human("Alice", 20)

users = [user1, user2, user3]

for user in users:
    user.check_adult()
```
